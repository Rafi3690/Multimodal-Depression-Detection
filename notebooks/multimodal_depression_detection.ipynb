```python
import torch
from PIL import Image
from tqdm import tqdm
from transformers import ViTModel, ViTImageProcessor
import numpy as np

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = ViTModel.from_pretrained("google/vit-base-patch16-224").to(device)
model.eval()
processor = ViTImageProcessor.from_pretrained("google/vit-base-patch16-224")

features = []

for path in tqdm(image_df["image_path"], desc="Extracting ViT features"):
    try:
        image = Image.open(path)
        if image.mode == "L":
            image = np.stack([np.array(image)]*3, axis=-1)
            image = Image.fromarray(image)
    except Exception as e:
        print(f"Skipping {path}: {e}")
        continue

    inputs = processor(images=image, return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)
        pooled = outputs.pooler_output
        features.append(pooled.cpu())

embeddings = torch.cat(features, dim=0)
print("Embeddings shape:", embeddings.shape)
```

```python
def remove_duration_outliers(df, duration_col='duration'):
    Q1 = df[duration_col].quantile(0.25)
    Q3 = df[duration_col].quantile(0.85)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    cleaned_df = df[(df[duration_col] >= lower_bound) &
                   (df[duration_col] <= upper_bound)]
    return cleaned_df
audio_df = remove_duration_outliers(audio_df)
plt.figure(figsize=(7, 4))
sns.histplot(audio_df['duration'], kde=True, bins=30)
plt.title('Distribution of Audio Durations (Outliers Removed)')
plt.xlabel('Duration (seconds)')
plt.ylabel('Count')
plt.show()
```

```python
from transformers import BertModel, BertTokenizer, ViTFeatureExtractor, ViTModel, Wav2Vec2Processor, Wav2Vec2Model
import torch

text_processor = BertTokenizer.from_pretrained("google-bert/bert-base-uncased")
text_model = BertModel.from_pretrained("google-bert/bert-base-uncased")

image_processor = ViTFeatureExtractor.from_pretrained("google/vit-base-patch16-224")
image_model = ViTModel.from_pretrained("google/vit-base-patch16-224")

audio_processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
audio_model = Wav2Vec2Model.from_pretrained("facebook/wav2vec2-base-960h")
```

```python
import torch
from tqdm import tqdm

batch_size = 16
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
text_model = text_model.to(device)
def extract_text_features(text_list):
    features = []
    for i in tqdm(range(0, len(text_list), batch_size)):
        batch = text_list[i:i + batch_size]
        inputs = text_processor(batch, padding=True, truncation=True, return_tensors="pt")
        inputs = {key: val.to(device) for key, val in inputs.items()}
        with torch.no_grad():
            outputs = text_model(**inputs)
        last_hidden_state = outputs.last_hidden_state
        batch_features = last_hidden_state.mean(dim=1)
        features.append(batch_features)
    return torch.cat(features, dim=0)

train_input = train_dataset['text'].tolist()
val_input = val_dataset['text'].tolist()
test_input = test_dataset['text'].tolist()

train_text_features = extract_text_features(train_input)
val_text_features = extract_text_features(val_input)
test_text_features = extract_text_features(test_input)

print("Train:", train_text_features.shape)
print("Validation:", val_text_features.shape)
print("Test:", test_text_features.shape)
```

```python
mean = train_text_features.mean(dim=0, keepdim=True)
std = train_text_features.std(dim=0, keepdim=True)

train_text_features = (train_text_features - mean) / std
val_text_features = (val_text_features - mean) / std
test_text_features = (test_text_features - mean) / std
```

```python

def extract_image_features(image_paths, model, processor, batch_size=16):
    model.eval()
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    image_model = model.to(device)
    features = []

    for i in tqdm(range(0, len(image_paths), batch_size)):
        batch_paths = image_paths[i:i+batch_size]
        images = [Image.open(p).convert("RGB").resize((224, 224), Image.BICUBIC) for p in batch_paths]
        inputs = processor(images, return_tensors="pt", padding=True, truncation=True)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        with torch.no_grad():
            outputs = model(**inputs)
        batch_feats = outputs.last_hidden_state.mean(dim=1)
        features.append(batch_feats)

    return torch.cat(features, dim=0)


train_image_features = extract_image_features(train_dataset['image_path'].tolist(), image_model, image_processor)
val_image_features = extract_image_features(val_dataset['image_path'].tolist(), image_model, image_processor)
test_image_features = extract_image_features(test_dataset['image_path'].tolist(), image_model, image_processor)

print("Train:", train_image_features.shape)
print("Validation:", val_image_features.shape)
print("Test:", test_image_features.shape)
```

```python
import torchaudio
import torch
from tqdm import tqdm

def extract_audio_features(audio_paths, model, processor, batch_size=16):
    model.eval()
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    features = []
    for i in tqdm(range(0, len(audio_paths), batch_size)):
        batch_paths = audio_paths[i:i+batch_size]
        waveforms = []
        for path in batch_paths:
            waveform, sr = torchaudio.load(path)
            if sr != 16000:
                resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
                waveform = resampler(waveform)
            if waveform.shape[0] > 1:
                waveform = waveform.mean(dim=0, keepdim=True)
            waveforms.append(waveform.squeeze(0).numpy())
        inputs = processor( waveforms, return_tensors="pt",sampling_rate=16000, padding=True )
        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = model(**inputs)
        batch_feats = outputs.last_hidden_state.mean(dim=1)  # (batch, hidden)
        features.append(batch_feats)
    return torch.cat(features, dim=0).to(device)

train_audio_features = extract_audio_features(train_dataset['audio_path'].tolist(), audio_model, audio_processor)
val_audio_features   = extract_audio_features(val_dataset['audio_path'].tolist(), audio_model, audio_processor)
test_audio_features   = extract_audio_features(test_dataset['audio_path'].tolist(), audio_model, audio_processor)

print("Train:", train_audio_features.shape)
print("Validation:", val_audio_features.shape)
print("Test:", test_audio_features.shape)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class MultiModalFusionModel(nn.Module):
    def __init__(self, feature_dim=768, num_heads=8, hidden_dim=512,output_dim=4, num_layers=12, dropout=0.3, modality_dropout=0.2, attn_heads=6):
        super().__init__()
        self.feature_dim = feature_dim
        self.num_heads = num_heads
        self.modality_dropout = modality_dropout
        self.attn_heads = attn_heads

        def proj():
            return nn.Linear(feature_dim, feature_dim)
        self.q_text, self.k_text, self.v_text = proj(), proj(), proj()
        self.q_audio, self.k_audio, self.v_audio = proj(), proj(), proj()
        self.q_image, self.k_image, self.v_image = proj(), proj(), proj()

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=feature_dim, nhead=num_heads,
            batch_first=True, dropout=dropout, activation="gelu"
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        self.attn_pool = nn.Linear(feature_dim, attn_heads)

        def gate_block():
            return nn.Sequential(
                nn.Linear(2 * feature_dim, feature_dim),
                nn.ReLU(),
                nn.Dropout(dropout),
                nn.Linear(feature_dim, feature_dim),
                nn.Sigmoid()
            )

        self.gate_text = gate_block()
        self.gate_audio = gate_block()
        self.gate_image = gate_block()

        self.norm = nn.LayerNorm(feature_dim)

        self.classifier = nn.Sequential(
            nn.Linear(feature_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim // 2, output_dim)
        )

    def forward(self, f_text, f_audio, f_image):
        B = f_text.size(0)

        if self.training:
            if torch.rand(1).item() < self.modality_dropout:
                f_text = torch.zeros_like(f_text)
            if torch.rand(1).item() < self.modality_dropout:
                f_audio = torch.zeros_like(f_audio)
            if torch.rand(1).item() < self.modality_dropout:
                f_image = torch.zeros_like(f_image)

        q_t, k_t, v_t = self.q_text(f_text), self.k_text(f_text), self.v_text(f_text)
        q_a, k_a, v_a = self.q_audio(f_audio), self.k_audio(f_audio), self.v_audio(f_audio)
        q_v, k_v, v_v = self.q_image(f_image), self.k_image(f_image), self.v_image(f_image)

        seq = torch.stack([q_t, q_a, q_v], dim=1)
        fused_seq = self.transformer(seq)

        attn_weights = torch.softmax(self.attn_pool(fused_seq), dim=1)
        attn_weights = attn_weights.permute(0, 2, 1)
        pooled = torch.einsum("bhc,bcd->bhd", attn_weights, fused_seq)
        mmf = pooled.mean(dim=1)

        g_t = self.gate_text(torch.cat([f_text, mmf], dim=-1))
        g_a = self.gate_audio(torch.cat([f_audio, mmf], dim=-1))
        g_v = self.gate_image(torch.cat([f_image, mmf], dim=-1))

        gated_fusion = mmf + (g_t * f_text + g_a * f_audio + g_v * f_image)
        gated_fusion = self.norm(gated_fusion)

        logits = self.classifier(gated_fusion)

        return logits, {
            "gates": torch.cat(
                [g_t.mean(-1, keepdim=True),
                 g_a.mean(-1, keepdim=True),
                 g_v.mean(-1, keepdim=True)], dim=-1
            )
        }
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score
import numpy as np

class MultiModalDataset(Dataset):
    def __init__(self, text_feats, img_feats, audio_feats, labels):
        self.text_feats = text_feats
        self.img_feats = img_feats
        self.audio_feats = audio_feats
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            "text": self.text_feats[idx],
            "image": self.img_feats[idx],
            "audio": self.audio_feats[idx],
            "label": self.labels[idx]
        }
```

```python
train_dataset = MultiModalDataset(train_text_features, train_image_features, train_audio_features, train_labels)
val_dataset   = MultiModalDataset(val_text_features, val_image_features, val_audio_features, val_labels)
test_dataset   = MultiModalDataset(test_text_features, test_image_features, test_audio_features, test_labels)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=16, shuffle=False)
test_loader   = DataLoader(test_dataset, batch_size=16, shuffle=False)
```

```python
import torch
from tqdm import tqdm
from sklearn.metrics import accuracy_score
import torch_optimizer as optim

optimizer = torch.optim.RAdam(model.parameters(), lr=5e-4, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2, eta_min=1e-4)
criterion = torch.nn.CrossEntropyLoss(label_smoothing=0.04)
torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
```

```python
from tqdm import tqdm
from sklearn.metrics import accuracy_score

history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_val_loss = float("inf")
patience, patience_counter = 10, 12

epochs = 200
for epoch in range(epochs):
    model.train()
    train_loss = 0.0
    all_preds, all_labels = [], []

    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs} [Train]"):
        f_text = batch["text"].to(device)
        f_img = batch["image"].to(device)
        f_audio = batch["audio"].to(device)
        labels = batch["label"].to(device)

        optimizer.zero_grad()
        logits, _ = model(f_text, f_audio, f_img)
        loss = criterion(logits, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        preds = torch.argmax(logits, dim=1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    avg_train_loss = train_loss / len(train_loader)
    train_acc = accuracy_score(all_labels, all_preds)

    model.eval()
    val_loss = 0.0
    val_preds, val_labels = [], []
    with torch.no_grad():
        for batch in tqdm(val_loader, desc=f"Epoch {epoch+1}/{epochs} [Val]"):
            f_text = batch["text"].to(device)
            f_img = batch["image"].to(device)
            f_audio = batch["audio"].to(device)
            labels = batch["label"].to(device)

            logits, _ = model(f_text, f_audio, f_img)
            loss = criterion(logits, labels)

            val_loss += loss.item()
            preds = torch.argmax(logits, dim=1)
            val_preds.extend(preds.cpu().numpy())
            val_labels.extend(labels.cpu().numpy())

    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_labels, val_preds)

    scheduler.step(avg_val_loss)

    # if avg_val_loss < best_val_loss:
    #     best_val_loss = avg_val_loss
    #     patience_counter = 0
    #     torch.save(model.state_dict(), "best_model.pth")
    #     print("  Model saved (val loss improved).")
    # else:
    #     patience_counter += 1
    #     if patience_counter >= patience:
    #         print("  Early stopping triggered!")
    #         break

    history["train_loss"].append(avg_train_loss)
    history["val_loss"].append(avg_val_loss)
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)

    print(f"Epoch [{epoch+1}/{epochs}] "
          f"LR: {optimizer.param_groups[0]['lr']:.6f} "
          f"Train Loss: {avg_train_loss:.4f}, Train Acc: {train_acc:.4f} "
          f"Val Loss: {avg_val_loss:.4f}, Val Acc: {val_acc:.4f}")
```

```python
import matplotlib.pyplot as plt
def plot_history(history):
    epochs = range(1, len(history["train_loss"]) + 1)
    plt.figure(figsize=(12,5))
    plt.subplot(1,2,1)
    plt.plot(epochs, history["train_loss"], label="Train Loss", marker="o")
    plt.plot(epochs, history["val_loss"], label="Val Loss", marker="o")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.title("Training vs Validation Loss")
    plt.legend()
    plt.grid(True)

    plt.subplot(1,2,2)
    plt.plot(epochs, history["train_acc"], label="Train Accuracy", marker="o")
    plt.plot(epochs, history["val_acc"], label="Val Accuracy", marker="o")
    plt.xlabel("Epochs")
    plt.ylabel("Accuracy")
    plt.title("Training vs Validation Accuracy")
    plt.legend()
    plt.grid(True)
    plt.show()
plot_history(history)
```

```python
from sklearn.metrics import classification_report

def evaluate_model(model, data_loader, device="cuda", target_names=None):
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for batch in data_loader:
            f_text = batch["text"].to(device)
            f_img = batch["image"].to(device)
            f_audio = batch["audio"].to(device)
            labels = batch["label"].to(device)

            logits, _ = model(f_text, f_audio, f_img)
            preds = torch.argmax(logits, dim=1)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    print("\n Classification Report:")
    print(classification_report(all_labels, all_preds, target_names=target_names))

target_names = ["Highly Positive", "Positive", "Negative", "Neutral"]
evaluate_model(model, val_loader, device=device, target_names=target_names)
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import numpy as np

def plot_pr_curve(y_true, y_probs, n_classes):
    y_true_bin = label_binarize(y_true, classes=np.arange(n_classes))

    precision = dict()
    recall = dict()
    avg_precision = dict()

    for i in range(n_classes):
        precision[i], recall[i], _ = precision_recall_curve(y_true_bin[:, i], y_probs[:, i])
        avg_precision[i] = average_precision_score(y_true_bin[:, i], y_probs[:, i])

    precision["micro"], recall["micro"], _ = precision_recall_curve(
        y_true_bin.ravel(), y_probs.ravel()
    )
    avg_precision["micro"] = average_precision_score(y_true_bin, y_probs, average="micro")

    plt.figure(figsize=(8,6))
    for i in range(n_classes):
        plt.plot(recall[i], precision[i],
                 label=f"Class {i} (AP = {avg_precision[i]:.2f})")

    plt.plot(recall["micro"], precision["micro"],
             label=f"Micro-average (AP = {avg_precision['micro']:.2f})",
             color="navy", linestyle="--", linewidth=2)

    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve")
    plt.legend(loc="lower left")
    plt.grid(True)
    plt.show()
```
