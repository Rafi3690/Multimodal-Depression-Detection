```python
def remove_duration_outliers(df, duration_col='duration'):
    Q1 = df[duration_col].quantile(0.25)
    Q3 = df[duration_col].quantile(0.85)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    cleaned_df = df[(df[duration_col] >= lower_bound) & 
                   (df[duration_col] <= upper_bound)]
    return cleaned_df

audio_df = remove_duration_outliers(audio_df)

plt.figure(figsize=(7, 4))
sns.histplot(audio_df['duration'], kde=True, bins=30)
plt.title('Distribution of Audio Durations (Outliers Removed)')
plt.xlabel('Duration (seconds)')
plt.ylabel('Count')
plt.show()
```

Audio preprocessing

```python
import librosa
import pandas as pd
import soundfile as sf
import os
import zipfile
import noisereduce as nr
import numpy as np

processed_audio_dir = 'processed_audio_files'
os.makedirs(processed_audio_dir, exist_ok=True)
processed_audio_indexed_paths = []
def is_valid_audio(audio):
    return np.all(np.isfinite(audio)) and not np.isnan(audio).any()
for i, row in multimodal_df.iterrows():
    audio_path = row['audio_path']
    try:
        if not os.path.exists(audio_path):
            print(f" Audio file not found: {audio_path}")
            continue
        audio, sr = librosa.load(audio_path, sr=16000)
        shifted_audio = librosa.effects.pitch_shift(audio, sr=sr, n_steps=2)
        denoised_audio = nr.reduce_noise(y=shifted_audio, sr=sr)
        y_harmonic, y_percussive = librosa.effects.hpss(denoised_audio)
        harmonic = librosa.effects.time_stretch(y_harmonic, rate=0.9)
        trimmed_audio, _ = librosa.effects.trim(harmonic)

        if not is_valid_audio(trimmed_audio):
            print(f" Skipping index {i} — NaN or Inf detected after processing")
            continue

        processed_audio_path = os.path.join(processed_audio_dir, f'processed_audio_{i}.wav')
        sf.write(processed_audio_path, trimmed_audio, sr)
        processed_audio_indexed_paths.append((i, processed_audio_path))

    except Exception as e:
        print(f" Error processing audio at index {i}: {e}")

multimodal_df['processed_audio_path'] = None  
for idx, path in processed_audio_indexed_paths:
    multimodal_df.loc[idx, 'processed_audio_path'] = path

print(f" Updated multimodal_df with {len(processed_audio_indexed_paths)} processed audio entries.")
```

```python
import torch

batch_size = 8  
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
text_model = text_model.to(device)
def extract_text_features(text_list):
    features = []
    for i in range(0, len(text_list), batch_size):
        batch = text_list[i:i + batch_size]
        inputs = text_processor(batch, padding=True, truncation=True, return_tensors="pt")
        inputs = {key: val.to(device) for key, val in inputs.items()}
        with torch.no_grad():
            outputs = text_model(**inputs)
        last_hidden_state = outputs.last_hidden_state
        batch_features = last_hidden_state.mean(dim=1)
        features.append(batch_features)
    
    return torch.cat(features, dim=0)

train_input = train_dataset['text'].tolist()
val_input = val_dataset['text'].tolist()

train_text_features = extract_text_features(train_input)
val_text_features = extract_text_features(val_input)

print("Train:", train_text_features.shape)
print("Validation:", val_text_features.shape)
```

Image feature extract

```python
def extract_image_features(image_paths, model, processor, batch_size=8):
    model.eval()
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    features = []

    for i in range(0, len(image_paths), batch_size):
        batch_paths = image_paths[i:i+batch_size]
        images = [Image.open(p).convert("RGB").resize((224, 224), Image.BICUBIC) for p in batch_paths]
        inputs = processor(images, return_tensors="pt", padding=True, truncation=True)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        with torch.no_grad():
            outputs = model(**inputs)
        batch_feats = outputs.last_hidden_state.mean(dim=1)
        features.append(batch_feats)

    return torch.cat(features, dim=0)


train_image_features = extract_image_features(train_dataset['image_path'].tolist(), image_model, image_processor)
val_image_features = extract_image_features(val_dataset['image_path'].tolist(), image_model, image_processor)

print("Train:", train_image_features.shape)
print("Validation:", val_image_features.shape)
```

Audio features Extract

```python
import torchaudio
def extract_audio_features(audio_paths, model, processor):
    model.eval()
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model = model.to(device)
    features = []

    for path in audio_paths:
        waveform, sr = torchaudio.load(path)
        if sr != 16000:
            resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
            waveform = resampler(waveform)
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0, keepdim=True)

        inputs = processor(waveform.squeeze(0), return_tensors="pt", sampling_rate=16000, padding=True)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        with torch.no_grad():
            outputs = model(**inputs)
        features.append(outputs.last_hidden_state.mean(dim=1))

    return torch.cat(features, dim=0)
train_audio_paths = train_dataset['audio_path'].tolist()
val_audio_paths = val_dataset['audio_path'].tolist()

train_audio_features = extract_audio_features(train_audio_paths, audio_model, audio_processor)
val_audio_features = extract_audio_features(val_audio_paths, audio_model, audio_processor)

print("Train:", train_audio_features.shape)
print("Validation:", val_audio_features.shape)
```

```python
mean = train_audio_features.mean(dim=0, keepdim=True)
std = train_audio_features.std(dim=0, keepdim=True)

train_audio_features = (train_audio_features - mean) / std
val_audio_features = (val_audio_features - mean) / std
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class EarlyFusion(nn.Module):
    def __init__(self, input_dim, num_classes):
        super(EarlyFusion, self).__init__()
        self.fc1 = nn.Linear(input_dim, 1024)
        self.bn1 = nn.BatchNorm1d(1024)
        self.fc2 = nn.Linear(1024, 512)
        self.bn2 = nn.BatchNorm1d(512)
        self.fc3 = nn.Linear(512, 256)
        self.bn3 = nn.BatchNorm1d(256)
        self.fc4 = nn.Linear(256, num_classes)

        self.dropout = nn.Dropout(0.4)
        self.gelu = nn.GELU()

    def forward(self, x):
        x = self.gelu(self.bn1(self.fc1(x)))
        x = self.dropout(x)
        x = self.gelu(self.bn2(self.fc2(x)))
        x = self.dropout(x)
        x = self.gelu(self.bn3(self.fc3(x)))
        x = self.dropout(x)
        x = self.fc4(x)
        return x
```

```python
import torch.nn as nn

class LateFusionModel(nn.Module):
    def __init__(self, input_dim_text, input_dim_image, input_dim_audio, num_classes):
        super(LateFusionModel, self).__init__()
        fusion_dim = input_dim_text + input_dim_image + input_dim_audio
        self.fc = nn.Sequential(
            nn.Linear(fusion_dim, 256),
            nn.GELU(),
            nn.Dropout(0.4),
            nn.Linear(256, num_classes)
        )

    def forward(self, text_features, image_features, audio_features):
        fused = torch.cat((text_features, image_features, audio_features), dim=1)
        return self.fc(fused)
```

Hybrid Fusion

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class HybridFusionModel(nn.Module):
    def __init__(self, input_dim_text, input_dim_image, input_dim_audio, hidden_dim=512, num_classes=4):
        super(HybridFusionModel, self).__init__()
        self.text_branch = nn.Sequential(
            nn.Linear(input_dim_text, hidden_dim),
            nn.GELU(),
            nn.Dropout(0.4)
        )
        self.image_branch = nn.Sequential(
            nn.Linear(input_dim_image, hidden_dim),
            nn.GELU(),
            nn.Dropout(0.5)
        )
        self.audio_branch = nn.Sequential(
            nn.Linear(input_dim_audio, hidden_dim),
            nn.GELU(),
            nn.Dropout(0.3)
        )
        self.fusion = nn.Sequential(
            nn.Linear(hidden_dim * 3, hidden_dim),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(hidden_dim, num_classes)
        )

    def forward(self, text_feat, image_feat, audio_feat):
        text_out = self.text_branch(text_feat)
        image_out = self.image_branch(image_feat)
        audio_out = self.audio_branch(audio_feat)
        fused = torch.cat([text_out, image_out, audio_out], dim=1)
        output = self.fusion(fused)
        return output
```

```python
input_dim_text = train_text_features.shape[1]
input_dim_image = train_image_features.shape[1]
input_dim_audio = train_audio_features.shape[1]
num_classes = 4
hidden_dim = 256

model = HybridFusionModel(input_dim_text, input_dim_image, input_dim_audio, hidden_dim, num_classes).to(device)
model
```

```python
model = HybridFusionModel(input_dim_text, input_dim_image, input_dim_audio, num_classes=4).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=10e-4,weight_decay=0.03)
```

```python
train_losses, val_losses = [], []
train_accuracies, val_accuracies = [], []
for epoch in range(130):
    model.train()
    optimizer.zero_grad()
    outputs = model(train_text_features, train_image_features, train_audio_features)
    loss = criterion(outputs, train_labels)
    _, predicted = torch.max(outputs, 1)
    acc = (predicted == train_labels).float().mean()
    loss.backward()
    optimizer.step()
    train_losses.append(loss.item())
    train_accuracies.append(acc.item())
    model.eval()
    with torch.no_grad():
        val_outputs = model(val_text_features, val_image_features, val_audio_features)
        val_loss = criterion(val_outputs, val_labels)
        _, val_predicted = torch.max(val_outputs, 1)
        val_acc = (val_predicted == val_labels).float().mean()
    val_losses.append(val_loss.item())
    val_accuracies.append(val_acc.item())

    print(f"Epoch {epoch+1}: Train Loss={loss.item():.2f}, Acc={acc.item():.2f} | Val Loss={val_loss.item():.2f}, Acc={val_acc.item():.2f}")
```

```python
from sklearn.metrics import classification_report
model.eval()
with torch.no_grad():
    val_outputs = model(val_text_features, val_image_features, val_audio_features)
    _, val_preds = torch.max(val_outputs, 1)

val_preds = val_preds.cpu()
val_labels_cpu = val_labels.cpu()
target_names = ['Highly Positive', 'Positive', 'Negative', 'Neutral']
report = classification_report(val_labels_cpu, val_preds, target_names=target_names)
print(report)
```
